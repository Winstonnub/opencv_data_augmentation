# Data Augmentation Examples

```python
import cv2 #import OpenCV 
import numpy as np #import numpy for array manipulations
import os #import os for file path
```

```python
folder_path = "./pics" # Path to input folder
save_path = "./output" # Path to save augmented images

# Perform augmentation for each image in the folder
for filename in os.listdir(folder_path): # Loop through each folder
    if filename.endswith(".jpg") or filename.endswith(".png"): # Check for file type
        # Load image
        img_path = os.path.join(folder_path, filename) # Get full image path
        img = cv2.imread(img_path) # Read image using OpenCV
        cols = img.shape[1] # Get number of columns in the image
        rows = img.shape[0] # Get number of rows in the image
        
        # Apply basic augmentation techniques to create new images
        flipped_img = cv2.flip(img, 1) # Flip image horizontally
        rotated_img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE) # Rotate image 90 degrees clockwise
        gaussian_blur_img = cv2.GaussianBlur(img, (5,5), 0) # Apply Gaussian blur
        gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY ) # Convert to grayscale
        color_img = cv2.cvtColor(img, cv2.COLOR_BGR2HSV ) # Convert to HSV color space (Hue, Saturation and Value)
        

        # Geometric Transformation of Images
        # Resizing with interpolation
        larger_img = cv2.resize(img, None, fx=2, fy=2, interpolation = cv2.INTER_LINEAR) # Resize image to larger size, using interpolation, (data source, image sizwe, fx, fy, interpolation method)
        # Translation
        translate_img = cv2.warpAffine(img, np.float32([[1, 0, 100], [0, 1, 50]]), (cols, rows)) # Translate image by (100, 50) pixels
        # Rotation
        M = cv2.getRotationMatrix2D((cols/2, rows/2), 45, 1) # Get rotation matrix for rotating image by 45 degrees around its center
        rotated_img = cv2.warpAffine(img, M, (cols, rows)) # Rotate image using the rotation matrix
        

        # Smoothing Images
        # Averaging Blurring
        averaging_blur_img = cv2.blur(img, (5,5)) # Apply averaging blur with a 5x5 kernel. # You look at small neighbourhood of each pixel and take average value of all pixels in that neighbourhood and assign it to the central pixel
        
        # Morphological Transformations
        # Erosion (thin out boundaries of foreground object)
        kernel = np.ones((5,5), np.uint8) # Create a 5x5 kernel of ones (central pixel is the pixel we looking at)
        erosion_img = cv2.erode(img, kernel, iterations=1) # Erode image # Erosion removes pixels on object boundaries, the pixel will be eroded if not all pixels under the kernel is 1
        # Dilation (thicken boundaries of foreground object)
        dilation_img = cv2.dilate(img,kernel,iterations = 1) # Dilate image # Dilation adds pixels to the boundaries of objects in an image, the pixel will be dilated if at least one pixel under the kernel is 1
        # Opening (erosion followed by dilation) - useful for removing noise, works as like, if there are some small dots outside of the main object, they will be removed, then we thicken the main object back to its original size
        opening_img = cv2.morphologyEx(img, cv2.MORPH_OPEN, kernel)
        # Closing (dilation followed by erosion) - useful for closing small holes inside the foreground objects, works as like, if there are some small black dots inside the main object, they will be removed, then we thin the main object back to its original size
        closing_img = cv2.morphologyEx(img, cv2.MORPH_CLOSE, kernel)
        # Morphological Gradient - OUTLINE of the object (difference between dilation and erosion) - Basically it is used to get the outline of the object
        gradient_img = cv2.morphologyEx(img, cv2.MORPH_GRADIENT, kernel)


        # Structuring ELements (get your own shaped kernel!)
        kernel_rec = cv2.getStructuringElement(cv2.MORPH_RECT,(5,5))
        kernel_ell = cv2.getStructuringElement(cv2.MORPH_ELLIPSE,(5,5))
        kernel_cross = cv2.getStructuringElement(cv2.MORPH_CROSS,(5,5))
        # You can use these kernels in the morphological transformations above to get different effects


        # Edge Detection
        # Canny Edge Detection
        edges_img = cv2.Canny(img,100,200) # Apply Canny edge detection, intensity gradients are used to find edges in the image, # The first threshold is for the lower boundary, the second threshold is for the upper boundary

        # Save new images
        new_filename_1 = "01flipped_" + filename
        new_filename_2 = "02rotated_" + filename
        new_filename_3 = "03blurred_" + filename
        new_filename_4 = "04grayed_" + filename
        new_filename_5 = "05colored_" + filename
        new_filename_6 = "06larger_" + filename
        new_filename_7 = "07translated_" + filename
        new_filename_8 = "08rotated2_" + filename
        new_filename_9 = "09averaging_blurred_" + filename
        new_filename_10 = "10eroded_" + filename
        new_filename_11 = "11dilated_" + filename
        new_filename_12 = "12opened_" + filename
        new_filename_13 = "13closed_" + filename
        new_filename_14 = "14gradient_" + filename
        new_filename_15 = "15edges_" + filename



        


        
        new_img_path_1 = os.path.join(save_path, new_filename_1)
        new_img_path_2 = os.path.join(save_path, new_filename_2)
        new_img_path_3 = os.path.join(save_path, new_filename_3)
        new_img_path_4 = os.path.join(save_path, new_filename_4)
        new_img_path_5 = os.path.join(save_path, new_filename_5)
        new_img_path_6 = os.path.join(save_path, new_filename_6)
        new_img_path_7 = os.path.join(save_path, new_filename_7)
        new_img_path_8 = os.path.join(save_path, new_filename_8)
        new_img_path_9 = os.path.join(save_path, new_filename_9)
        new_img_path_10 = os.path.join(save_path, new_filename_10)
        new_img_path_11 = os.path.join(save_path, new_filename_11)
        new_img_path_12 = os.path.join(save_path, new_filename_12)
        new_img_path_13 = os.path.join(save_path, new_filename_13)
        new_img_path_14 = os.path.join(save_path, new_filename_14)
        new_img_path_15 = os.path.join(save_path, new_filename_15)


        cv2.imwrite(new_img_path_1, flipped_img)
        cv2.imwrite(new_img_path_2, rotated_img)
        cv2.imwrite(new_img_path_3, gaussian_blur_img)
        cv2.imwrite(new_img_path_4, gray_img)
        cv2.imwrite(new_img_path_5, color_img)
        cv2.imwrite(new_img_path_6, larger_img)
        cv2.imwrite(new_img_path_7, translate_img)
        cv2.imwrite(new_img_path_8, rotated_img)
        cv2.imwrite(new_img_path_9, averaging_blur_img)
        cv2.imwrite(new_img_path_10, erosion_img)
        cv2.imwrite(new_img_path_11, dilation_img)
        cv2.imwrite(new_img_path_12, opening_img)
        cv2.imwrite(new_img_path_13, closing_img)
        cv2.imwrite(new_img_path_14, gradient_img)
        cv2.imwrite(new_img_path_15, edges_img)
```
